## 1. Spiking Neural Network (SNN) Predictor

```python
from snn_predictor import SNNPredictor, LeakyIntegrateAndFire

# Initialize the SNN
snn = SNNPredictor()

# Simulate predicting the next byte given a context
context = b"Hello, QRES!"
probs = snn.predict_next(context)

print(f"Context: {context}")
print(f"Predicted byte distribution (top 5):")
top5 = np.argsort(probs)[-5:][::-1]
for idx in top5:
    print(f"  Byte {idx} ('{chr(idx) if 32 <= idx < 127 else '?'}') -> P={probs[idx]:.4f}")
```

## 2. Quantum VQC Fusion

```python
from qnn_vqc import QNNPredictor

# Initialize the Quantum Circuit
qnn = QNNPredictor(n_qubits=4)

# Get entangled features from a data chunk
chunk = b"Quantum Supremacy Test Data..."
features = qnn.get_entangled_features(chunk)

print(f"Chunk: {chunk[:20]}...")
print(f"QNN Entangled Features (Qubit Probabilities): {features}")
print(f"Feature Sum (should be ~1.0 for valid state): {np.sum(features):.4f}")
```

## 3. Hive Mind (Federated Learning Simulation)

```python
from hive_mind import HiveMind

# Initialize the Hive Mind node
# Note: Requires metabrain_ppo_v5.zip to exist
try:
    hive = HiveMind(local_model_path="../ai/metabrain_ppo_v5.zip", swarm_id="demo_node")
    
    # Generate an Epiphany (shareable weights)
    epiphany = hive.generate_epiphany()
    print(f"Node ID: {epiphany['node_id']}")
    print(f"Fidelity Score: {epiphany['fidelity_score']}")
    print(f"Weight Keys: {list(epiphany['weights'].keys())}")
except FileNotFoundError as e:
    print(f"Hive Mind demo skipped: {e}")
```

## 4. End-to-End Compression (Optional)

Requires the `qres_rust` native module to be built.

```python
try:
    from qres import qres_rust
    
    data = b"The quick brown fox jumps over the lazy dog. " * 100
    compressed = qres_rust.encode_bytes(data, 0, b'')
    ratio = len(compressed) / len(data)
    
    print(f"Original: {len(data)} bytes")
    print(f"Compressed: {len(compressed)} bytes")
    print(f"Ratio: {ratio:.4f}")
except ImportError:
    print("qres_rust module not available. Run `maturin develop --release` first.")
```
